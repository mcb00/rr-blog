So what is causality anyway?
Well, we already have an intuitive grasp of causality developed from our experiences of daily life,
but, for doing science, we're going to need a fully fledged and logically consistent analytical framework for answering causal questions, i.e. the theory of *causal inference*.
But before we get there, you'll have to endulge me on a quick bit of historical context about how people were thinking about causality leading up to the development of causal inference.

## A brief history of causality

The roots of inquiry about causality extend deep into human history.
The Buddha weighed in on this about 2500 years ago with the [Idappaccayatā](https://en.wikipedia.org/wiki/Idappaccayat%C4%81):

> When this is, that is.
From the arising of this comes the arising of that.
When this isn't, that isn't.
From the cessation of this comes the cessation of that.
— the Buddha, circa 500 BCE

Notice that this expresses both association ("when this is, that is"), as well as the notion of directionality ("from the arising of this comes the arising of that.)

Much later in the 18th century, Scottish philosopher 
[David Hume](https://en.wikipedia.org/wiki/David_Hume)
took a stab at it:

> We may define a cause to be an object followed by another, and where all the objects, similar to the first, are followed by objects similar to the second. Or, in other words, where, if the first object had not been, the second never had existed.
— Hume, 1748

Notice that Hume is  expressing both association ("where all the objects, similar to the first, are followed by objects similar to the second") and also a notion of the *counterfactual* (if the first object had not been, the second never had existed").
*Counterfactual* means contrary to fact; it evokes the idea of a parallel universe where certain specific events are altered from their factual counterparts, allowing us to imagine what *would* happen in that case and compare it to what *actually* happened.
The connonical modern counterfactual conception of causality comes from philosopher David Lewis:

> We think of a cause as something that makes a difference, and the difference it makes must be a difference from what would have happened without it. Had it been absent, its effects – some of them, at least, and usually all – would have been absent as well.
— David Lewis, 1973

So we arrived at the idea that counterfactuals were important for understanding causality, but this did not yet help us to compute answers to causal questions.
