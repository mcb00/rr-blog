So what is causality anyway?
Well, we already have an intuitive grasp of causality developed from our experiences of daily life,
but, for doing science, we need more than common sense.
We're going to need a fully fledged and logically consistent analytical framework for answering quantitative
 causal questions; happily that framework does exist, and it is known as  *causal inference*.

 As with the field of statistics and its ideological rift between Bayesian and Frequentist methods, causal inference is also a field divided along methodological lines.
As with stat, in causal inference We find two traditions with two distinct intellectual lineages.
But unlike in stat, it is not always clear or obvious to which of these two traditions a particular method, analysis, or result belongs.
Worse, it is often very difficult or nearly impossible to translate ideas and assertions made in one tradition into the language of the other. 
Needless to say, this can make things very confusing indeed.
So, before we dive into the framework itself, it's worth taking a moment to understand the history of the development of the field and the sources of its two great lineages.

## A brief and very incomplete history of causality

### The Buddha and the philosophers

The roots of inquiry about causality extend deep into human history.
The Buddha articulated notions of cause and effect  about 2500 years ago with the [Idappaccayatā](https://en.wikipedia.org/wiki/Idappaccayat%C4%81):

> When this is, that is. \
From the arising of this comes the arising of that. \
When this isn't, that isn't.  
From the cessation of this comes the cessation of that. \
— the Buddha, circa 500 BCE

The Buddha wasn't trying to define causality, but notice that this passage points to two different causal-ish concepts.
The first and third lines express association — "when this is, that is" — two events that occur together.
To me, the second and fourth lines point to a specific directionality between the events, i.e. what happens with "this" impacts what happens with "that".


Much later in the 18th century, Scottish philosopher 
[David Hume](https://en.wikipedia.org/wiki/David_Hume)
attempted an actual definition of causality:

> We may define a cause to be an object followed by another, and where all the objects, similar to the first, are followed by objects similar to the second. Or, in other words, where, if the first object had not been, the second never had existed. \
— Hume, 1748

Humee is sort of double dipping here, defining causality in two different ways.
The first part expresses association — "where all the objects, similar to the first, are followed by objects similar to the second". 
The second part expresses a notion of the *counterfactual* — "if the first object had not been, the second never had existed" — i.e. if the first event hadn't happened, the second event wouldn't have happened.
Philosopher David Lewis developed this counterfactual notion of causality more fully in the 1970's:

> We think of a cause as something that makes a difference, and the difference it makes must be a difference from what would have happened without it. Had it been absent, its effects – some of them, at least, and usually all – would have been absent as well. \
— David Lewis, 1973

So from the Buddha, Hume, and Lewis, we have the following pieces: association between events, directionality of impact from cause to effect, , and the notion of counterfactual thinking about what would have happened if events were different than they actually were.
The philosophers brought the pieces to the table, but it would take scientists to assemble them into an analytical framework capable of answering quantitative causal questions.


## The Causal Hierarchy


When learning a new subject, it's often quite handy to have a sort of superstructure one can use to organize new information as it comes in; ideally, it helps illuminate where different ideas belong and how they fit together. The *causal hierarchy* will serve as that superstructure for us while learning causal inference. 
It is a hierarchy over the types of causal questions we can ask and the kinds of assumptions and data that we need to answer them.
There are three levels:

1. Associations
1. Interventions
1. Counterfactuals


It's a true hierarchy in the sense that answering questions at one level requires data and assumptions from either that level or a higher level.
Let's check out each level in turn.

### Level 1: Associations

Levvel 1 questions or queries are about the dependencies and associations among variables that we can see by observing the data generating process.
Examples of level 1 

** What proportion of gym members take creatine? **

** Of those taking creatine, how many also use whey protein? **

** What's the correlation between a person's height and max bench press weight?

These questions can be answered using observational data from the DGP alone; no modeling assumptions are needed.
They concern associations among variables that can be answered by estimating the conditional distribution of some variable(s) given the values of others.
Notably, there is no causation at this level.
